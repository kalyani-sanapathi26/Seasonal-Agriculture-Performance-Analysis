## 5. Data Quality / Missing Values

```python
# Identify missing values
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

print("Columns with missing values and their counts:")
print(missing_values)

# Calculate percentage of missing values
missing_percentage = (missing_values / len(df)) * 100
print("\nPercentage of missing values:")
print(missing_percentage)
```

### Assessment of Missing Values and Treatment Strategy

**Missing Values Identified:**

*   `Rainfall_mm`: 48 missing values (1.2%)
*   `Yield_Tonnes_Ha`: 32 missing values (0.8%)
*   `Soil_Moisture_pct`: 40 missing values (1.0%)

The percentage of missing values for these columns is very low (all below 2%). Given their importance to the analysis (Rainfall and Soil Moisture are key environmental factors, and Yield is a primary performance metric), removing rows with missing values would lead to a small but unnecessary loss of data.

**Proposed Treatment: Median Imputation**

For numerical variables with a small percentage of missing values, median imputation is a robust strategy. The median is less sensitive to outliers compared to the mean, which is beneficial as we haven't yet performed outlier analysis.

*   **`Rainfall_mm`:** Impute missing values with the median rainfall.
*   **`Yield_Tonnes_Ha`:** Impute missing values with the median yield per hectare.
*   **`Soil_Moisture_pct`:** Impute missing values with the median soil moisture percentage.

**Justification:**

*   **Low Percentage of Missingness:** Since the missing data constitutes a very small fraction of the total dataset, imputation is preferable to dropping rows, as it preserves the overall dataset size and statistical power.
*   **Nature of Variables:** `Rainfall_mm`, `Yield_Tonnes_Ha`, and `Soil_Moisture_pct` are continuous numerical variables where a central tendency measure like the median can provide a reasonable estimate for the missing data points without significantly distorting the distribution of these variables or introducing bias.
*   **Robustness:** Using the median ensures that potential outliers in the observed data do not unduly influence the imputed values.

```python
# Impute missing values with the median
df['Rainfall_mm'].fillna(df['Rainfall_mm'].median(), inplace=True)
df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median(), inplace=True)
df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median(), inplace=True)

print("Missing values after imputation:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nMissing value imputation complete for Rainfall_mm, Yield_Tonnes_Ha, and Soil_Moisture_pct.")
```

## 9. Outlier Analysis

### Investigating Potential Outliers

Outliers were visually observed in several numerical variables during the univariate analysis (e.g., `Rainfall_mm`, `Yield_Tonnes_Ha`, `Profit_INR`, `Water_Used_m3`). We will use the Interquartile Range (IQR) method to statistically identify outliers and then visualize them to understand their distribution better. For this analysis, we will focus on key numerical variables that showed visible outliers or are critical for performance analysis.

We will list the columns for which we want to check outliers and then calculate the lower and upper bounds for outlier detection using the IQR method. Outliers will be defined as values falling below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`.

```python
import numpy as np

# Columns identified as potentially having outliers or are critical for analysis
outlier_cols = [
    'Rainfall_mm',
    'Farm_Area_Hectares',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct',
    'Nitrogen_kg_ha',
    'Phosphorus_kg_ha',
    'Potassium_kg_ha',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Market_Price_INR_Tonne',
    'Total_Cost_INR',
    'Revenue_INR',
    'Profit_INR',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

# Store outlier counts and percentages
outlier_summary = {}

print("Outlier Analysis (using IQR method):\n")

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    num_outliers = len(outliers)
    percent_outliers = (num_outliers / len(df)) * 100

    outlier_summary[col] = {'count': num_outliers, 'percentage': percent_outliers}

    if num_outliers > 0:
        print(f"Column: {col}")
        print(f"  Number of outliers: {num_outliers} ({percent_outliers:.2f}%)")
        print(f"  Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")
        # Optional: display some outlier values
        # print(f"  Example outliers: {outliers[col].head().tolist()}\n")
    # else:
    #     print(f"Column: {col} - No outliers detected using IQR method.")

# Convert summary to DataFrame for better readability
outlier_df = pd.DataFrame.from_dict(outlier_summary, orient='index')
outlier_df.columns = ['Outlier Count', 'Outlier Percentage']
print("\nSummary of Outliers:")
display(outlier_df[outlier_df['Outlier Count'] > 0].sort_values(by='Outlier Count', ascending=False))
```

### Important Limitations of the Analysis

While this analysis provides valuable insights, it's essential to acknowledge its limitations:

1.  **Data Source and Scope:** The analysis is based on a single dataset, which may not represent the full diversity of agricultural practices or conditions across all regions or years. Generalizability might be limited.
2.  **Lack of Temporal Depth:** The dataset appears to be a snapshot without explicit time-series data (e.g., year-over-year performance), which limits the ability to analyze long-term trends or year-specific seasonal variations.
3.  **Linearity Assumption in Correlation:** Pearson correlation measures only linear relationships. More complex, non-linear interactions between variables might not have been fully captured.
4.  **Absence of Causal Inference:** While correlations and relationships were identified, this descriptive and exploratory analysis does not establish causation. Further statistical modeling (e.g., regression analysis, Granger causality) would be needed for causal claims.
5.  **Aggregated Data:** Some variables might be aggregated (e.g., `Avg_Temperature_C`). More granular data (e.g., daily temperature fluctuations, specific soil types beyond pH) could provide deeper insights.
6.  **Outlier Treatment:** The decision was made not to remove outliers, considering them legitimate observations for agricultural context. While this preserves data integrity, it means some statistical measures might be influenced by extreme values.
7.  **Implicit Crop Information:** While `Crop` is a categorical variable, the analysis did not delve into the specific biological or market characteristics of each crop beyond what's directly in the data. This could affect the interpretation of optimal conditions for different crops.
